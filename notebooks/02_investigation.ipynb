```python
import pandas as pd

games = pd.read_csv("../data/raw/games.csv")

games.head()
```

```python
#investigating and removing duplicates
```

```python
duplicate_games = games[games["GAME_ID"].duplicated(keep=False)]

duplicate_games.sort_values("GAME_ID").head(20)
```

```python
games["GAME_DATE_EST"] = pd.to_datetime(games["GAME_DATE_EST"])
```

```python
games["GAME_DATE_EST"].dtype
```

```python
games = games.sort_values("GAME_DATE_EST").reset_index(drop=True)
```

```python
games["TOTAL_POINTS"] = games["PTS_home"] + games["PTS_away"]
```

```python
games = games.drop_duplicates(subset="GAME_ID", keep="first")
```
